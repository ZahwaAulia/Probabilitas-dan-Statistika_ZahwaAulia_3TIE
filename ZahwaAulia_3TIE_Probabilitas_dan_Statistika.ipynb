```python
# Tabel Data
path = 'ZahwaAulia_3TIE.xlsx'
dataraw = pd.read_excel(path)
print(dataraw)
```

```python
#Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frequency")
print(datafrq)
```

```python
#Grafik Garis
datafrq.plot()
```

```python
#Grafik Batang
datafrq.plot(kind='bar')
```

```python
#Pie Chart
datafrq.plot(kind='pie',y='Frequency',autopct='%1.0f%%',legend=False)
```

```python
#Statistika Deskriptif
dataraw["Nilai"] = pd.to_numeric(dataraw["Nilai"],errors='coerce')
dt = dataraw["Nilai"]
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
print(stats)
```
